# ***`Phase 2: Data Preprocessing`***

## **`Data Cleaning`**

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import numpy as np
```

```python
# Check for missing values
print("Missing values per column:\n", df.isnull().sum())
```

```python
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1
outliers = ((df < (Q1 - 1.5 * IQR)) | (df > (Q3 + 1.5 * IQR))).sum()
print("\nNumber of potential outliers per column:\n", outliers)
```

```python
#  Fix Highly Skewed Features (Log Transformation)
skewed_cols = ['AveRooms', 'AveBedrms', 'AveOccup']
for col in skewed_cols:
    df[col] = np.log1p(df[col])
```

## ***` Feature Engineering `***

```python
df['RoomsPerHousehold'] = df['AveRooms'] / df['AveOccup']
df['BedroomsPerRoom'] = df['AveBedrms'] / df['AveRooms']
df['PopulationDensity'] = df['Population'] / df['AveOccup']
```

```python
#  Handle Correlated Features
#  Linear Regression: remove Longitude
df_lr = df.drop(columns=['Longitude'])

# Tree Models: keep all features
df_tree = df.copy()
```

## **`Train–Test Split`** 

```python
X_lr = df_lr.drop(columns=['MedHouseVal'])
y_lr = df_lr['MedHouseVal']

X_tree = df_tree.drop(columns=['MedHouseVal'])
y_tree = df_tree['MedHouseVal']

# Linear Regression dataset split
X_lr_train, X_lr_test, y_lr_train, y_lr_test = train_test_split(
    X_lr, y_lr, test_size=0.2, random_state=42
)

# Tree Models dataset split
X_tree_train, X_tree_test, y_tree_train, y_tree_test = train_test_split(
    X_tree, y_tree, test_size=0.2, random_state=42
)
```

```python
# Feature Scaling (only for Linear Regression)
scaler = StandardScaler()
X_lr_train_scaled = scaler.fit_transform(X_lr_train)
X_lr_test_scaled = scaler.transform(X_lr_test)

# Tree Models do not need scaling
X_tree_train_scaled = X_tree_train.values
X_tree_test_scaled = X_tree_test.values

# -----------------------------
print(" Preprocessing Complete")
# print(f"Linear Regression dataset shape: {X_lr_train_scaled.shape}")
# print(f"Tree Models dataset shape: {X_tree_train_scaled.shape}")
```

# ***`Phase 3: Model Development `***

## ***`Train Baseline Models`***

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
#  Linear Regression (requires scaled data)
lr_model = LinearRegression()
lr_model.fit(X_lr_train_scaled, y_lr_train);
```

```python
#  Decision Tree Regressor
dt_model = DecisionTreeRegressor(random_state=42)
dt_model.fit(X_tree_train_scaled, y_tree_train);
```

```python
#  Random Forest Regressor
rf_model = RandomForestRegressor(random_state=42, n_estimators=100)
rf_model.fit(X_tree_train_scaled, y_tree_train);
```

## ***`Model Evaluation `***

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

# -----------------------------
# Linear Regression predictions
y_lr_pred = lr_model.predict(X_lr_test_scaled)

# Decision Tree predictions
y_dt_pred = dt_model.predict(X_tree_test_scaled)

# Random Forest predictions
y_rf_pred = rf_model.predict(X_tree_test_scaled)

# -----------------------------
# Create evaluation function
def evaluate_model(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    return mae, mse, rmse, r2

# Evaluate all models
results = pd.DataFrame({
    'Model': ['Linear Regression', 'Decision Tree', 'Random Forest'],
    'MAE': [
        *evaluate_model(y_lr_test, y_lr_pred)[:1],
        *evaluate_model(y_tree_test, y_dt_pred)[:1],
        *evaluate_model(y_tree_test, y_rf_pred)[:1]
    ],
    'MSE': [
        *evaluate_model(y_lr_test, y_lr_pred)[1:2],
        *evaluate_model(y_tree_test, y_dt_pred)[1:2],
        *evaluate_model(y_tree_test, y_rf_pred)[1:2]
    ],
    'RMSE': [
        *evaluate_model(y_lr_test, y_lr_pred)[2:3],
        *evaluate_model(y_tree_test, y_dt_pred)[2:3],
        *evaluate_model(y_tree_test, y_rf_pred)[2:3]
    ],
    'R2 Score': [
        *evaluate_model(y_lr_test, y_lr_pred)[3:4],
        *evaluate_model(y_tree_test, y_dt_pred)[3:4],
        *evaluate_model(y_tree_test, y_rf_pred)[3:4]
    ]
})

print(" Model Evaluation Metrics:")
display(results.style.background_gradient(cmap='Blues'))
```

## ***`Model Interpretation `***

```python
# Function to get feature importance
def get_feature_importance(model, X_train):
    if hasattr(model, "feature_importances_"):
        importances = model.feature_importances_
        features = X_train.columns
        fi_df = pd.DataFrame({'Feature': features, 'Importance': importances}).sort_values(by='Importance', ascending=False)
        return fi_df
    else:
        print("Model does not have feature_importances_ attribute")
        return None

# Decision Tree Feature Importance
dt_fi = get_feature_importance(dt_model, X_tree_train)
print(" Decision Tree Feature Importance:")
display(dt_fi.style.background_gradient(cmap='Blues'))
```

```python

# Random Forest Feature Importance
rf_fi = get_feature_importance(rf_model, X_tree_train)
print(" Random Forest Feature Importance:")
display(rf_fi.style.background_gradient(cmap='Greens'))
```

# ***`Phase 4: Final Model & Deployment `***

```python
import plotly.express as px
import pandas as pd

# Melt the results DataFrame for plotting
metrics_plot = results.melt(id_vars='Model', 
                            value_vars=['MAE','MSE','RMSE','R2 Score'],
                            var_name='Metric', value_name='Score')

# Interactive bar plot using Plotly
fig = px.bar(metrics_plot, 
             x='Model', y='Score', color='Metric', barmode='group',
             title="Baseline Regression Models Comparison",
             text='Score')

fig.update_layout(yaxis_title="Score", xaxis_title="Model")
fig.show()
```

```python
"""The Random Forest Regressor is selected as the final model because it consistently outperforms both Linear Regression and Decision Tree across all evaluation metrics. It provides:
the lowest MAE, MSE, and RMSE, indicating more accurate predictions,
the highest R² score, showing strong explanatory power,
and strong robustness to nonlinear patterns, skewed features, and correlations.
Therefore, Random Forest is the optimal choice for predicting housing prices in the California Housing Dataset."""
```
